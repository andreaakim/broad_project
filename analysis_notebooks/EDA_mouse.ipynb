```python
import pandas as pd

#find NA
df = pd.read_csv("../data/mouse_files/gene_set_list_mouse_2024.txt", sep="\t", header=None, dtype=str)
gene_cols = df.columns[1:]
df[df[gene_cols].isna().any(axis=1)]
```

```python
#drop all rows with NA
df = pd.read_csv("../data/mouse_files/gene_set_list_mouse_2024.txt", sep="\t", header=None, dtype=str)
df_clean = df.dropna()
df_clean.rename(columns={0: "phenotype"}, inplace=True)
df_clean["genes"] = df_clean.iloc[:, 1:].agg("\t".join, axis=1)
df_clean = df_clean[["phenotype", "genes"]]

df_clean.head(15)

```

```python
df_clean.count()
```

```python
# find duplicate genes
gene_counts = df_clean["genes"].value_counts()

gene_counts.head(20)
```

```python
#get only phenotype names
df_clean = df.iloc[:, [0]]
df_clean
```

```python
#remove mp and replace _ with a space
df_clean.iloc[:, 0] = df_clean.iloc[:, 0].str.replace(r'^mp_', '', regex=True).str.replace(r'_', ' ', regex=True)
df_clean
```

```python
df_clean.to_csv("../data/mouse_files/gene_set_list_mouse_2024_clean.txt", sep="\t", index=False, header=False)
```
