# Used to look into the clusters to see what does and doesn't overlap

## Leiden

```python
from collections import defaultdict

def load_clusters(filepath):
    clusters = defaultdict(list)
    
    with open(filepath, "r") as f:
        for line in f:
            name, label = line.strip().split("\t")
            label = int(label.replace(",", ""))  
            clusters[label].append(name)
    
    return clusters
```

```python
clusters_lm = load_clusters("../data/mouse_files/minilm_leiden_clusters.txt")
clusters_pubmed = load_clusters("../data/mouse_files/pubmed_leiden_clusters.txt")

lm_set = set(clusters_lm[13])
pubmed_set = set(clusters_pubmed[12])

matching = lm_set & pubmed_set
lm_only = lm_set - pubmed_set
pubmed_only = pubmed_set - lm_set

print("Matching:", len(matching))
print("Only in LM:", len(lm_only))
print("Only in PubMed:", len(pubmed_only))
```

```python
with open("../data/mouse_files/lm_pm_comparison.txt", "w") as f:
    f.write("MATCHING:\n")
    for item in sorted(matching):
        f.write(item + "\n")
    
    f.write("\nONLY IN LM:\n")
    for item in sorted(lm_only):
        f.write(item + "\n")
    
    f.write("\nONLY IN PUBMED:\n")
    for item in sorted(pubmed_only):
        f.write(item + "\n")
```
